### Column Standardization

```python
df[['team1', 'team2', 'winner', 'toss_winner','city','venue']].nunique()
```

```python
df['venue'].unique()
```

```python
def standardize_venue(venue_name):
    if pd.isna(venue_name): 
        return venue_name
    
    if 'Feroz Shah Kotla' in venue_name or 'Arun Jaitley' in venue_name: 
        return 'Arun Jaitley Stadium'
    if 'Narendra Modi' in venue_name or 'Sardar Patel' in venue_name: 
        return 'Narendra Modi Stadium'
    
    if 'M.Chinnaswamy' in venue_name or 'M Chinnaswamy' in venue_name: 
        return 'M Chinnaswamy Stadium'
    if 'MA Chidambaram' in venue_name: 
        return 'MA Chidambaram Stadium'
    if 'Punjab Cricket Association' in venue_name: 
        return 'Punjab Cricket Association Stadium'
    if 'Rajiv Gandhi' in venue_name: 
        return 'Rajiv Gandhi International Stadium'
    return venue_name.split(',')[0].strip()

df['venue'] = df['venue'].apply(standardize_venue)
```

```python
all_teams = pd.concat([df['team1'], df['team2']]).dropna().unique()

print("Unique Teams:")
print(sorted(all_teams)) # Here Delhi Capitals and Delhi Daredevils is same team but with different name and so other teams too

print("\n\nUnique Cities:")
print(sorted(df['city'].dropna().unique())) # same here banglore and bengluru is the same city but with different names
```

```python
team_name_mapping = {
    'Rising Pune Supergiants': 'Rising Pune Supergiant',
    'Delhi Daredevils': 'Delhi Capitals',
    'Kings XI Punjab': 'Punjab Kings',
    'Royal Challengers Bangalore': 'Royal Challengers Bengaluru'
}

columns_to_update = ['team1', 'team2', 'winner', 'toss_winner']
for col in columns_to_update:
    df[col] = df[col].replace(team_name_mapping)
```

```python
df['city'] = df['city'].replace({'Bangalore': 'Bengaluru'})
```
